```python
products = {
    "Coffee X": {
        "price": "₹499",
        "category": "Premium Coffee",
        "description": "Premium roasted coffee beans"
    },
    "Coffee Y": {
        "price": "₹399",
        "category": "Classic Coffee",
        "description": "Medium roast coffee"
    }
}

offers = {
    "Coffee X": {
        "offer": "20% off"
    },
    "Coffee Y": {
        "offer": "10% off"
    }
}
def search_products(product_name):
    return products.get(product_name,"not found")

def get_product_offer(product_name):
    return offers.get(product_name,"no offer available")
print(search_products("Coffee X"))
print(get_product_offer("Coffee X"))
```
