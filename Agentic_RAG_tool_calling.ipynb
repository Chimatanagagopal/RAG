```python
from google.colab import userdata

gemini_api_key = userdata.get("GEMINI_API_KEY")

print("API key found:", gemini_api_key is not None)
```

```python
from google import genai

client = genai.Client(
    api_key=gemini_api_key
)

print("Gemini connected!")
```

```python
products = {
    "Coffee X": {
        "price": "₹499",
        "category": "Premium Coffee",
        "description": "Premium roasted coffee beans"
    },
    "Coffee Y": {
        "price": "₹399",
        "category": "Classic Coffee",
        "description": "Medium roast coffee"
    }
}

offers = {
    "Coffee X": {
        "offer": "20% off"
    },
    "Coffee Y": {
        "offer": "10% off"
    }
}


def search_product(product_name: str) -> dict:
    """Search product information by product name."""

    return products.get(
        product_name,
        {"error": "Product not found"}
    )


def search_offer(product_name: str) -> dict:
    """Search the current offer for a product."""

    return offers.get(
        product_name,
        {"error": "Offer not found"}
    )


print(search_product("Coffee X"))
print(search_offer("Coffee X"))
```

```python
from google.genai import types

config = types.GenerateContentConfig(
    tools=[
        search_product,
        search_offer
    ]
)
```

```python
response = client.models.generate_content(
    model="gemini-flash-latest",
    contents="What is the price of Coffee X?",
    config=config
)

print(response.text)
```
